## 4.2. <a id='toc4_2_'></a>[Pandas数值运算方法](#toc0_)

- NumPy 的基本能力之一是快速对每个元素进行运算，既包括基本算术运算（加、减、乘、
除），也包括更复杂的运算（三角函数、指数函数和对数函数等）。Pandas 继承了 NumPy
的功能

- 但是 Pandas 也实现了一些高效技巧：对于一元运算（像函数与三角函数），这些通用函
数将在输出结果中保留索引和列标签；而对于二元运算（如加法和乘法），Pandas 在传递
通用函数时会自动对齐索引进行计算。这就意味着，保存数据内容与组合不同来源的数
据——两处在 NumPy 数组中都容易出错的地方——变成了 Pandas 的杀手锏。后面还会介
绍一些关于一维 Series 和二维 DataFrame 的便捷运算方法。

## 4.9. <a id='toc4_9_'></a>[合并-concat/merge](#toc0_)


### 4.9.1. <a id='toc4_9_1_'></a>[Concat-轴(横向或纵向合并)](#toc0_)



* 沿着横 `横轴row axis=1向右` 或 `纵轴column axis=0向下` 进行 连接，而不是指定 `关键字(on='key')`；

* 常用：`pd.concat([df1, df2], axis=0, join='outer', ingore_index=False)`

* 关键参数：`axis、join`

参数说明：

| 参数                 |                  默认值 | 含义 / 逻辑                                                | 常见取值与效果                                      |
| ------------------ | -------------------: | ------------------------------------------------------ | -------------------------------------------- |
| `objs`             |                    — | 要合并的对象序列（list/tuple/dict），每个是 Series/DataFrame 或它们的可迭代 | `objs=[df1, df2, ...]` 或 `{'a':df1,'b':df2}` |
| `axis`             |                  `0` | 沿哪个轴拼接：`0` 按行（垂直，堆叠行），`1` 按列（水平，拼列）                    | `axis=0`：结果行数累加，列为并集；`axis=1`：列累加，行按索引对齐     |
| `join`             |            `'outer'` | 对齐方式（当索引/列不一致时）：`outer` 保留并集，缺失用 NaN；`inner` 取交集       | `join='outer'`（默认）或 `'inner'`                |
| `ignore_index`     |              `False` | 是否忽略原有索引，新索引使用连续整数 `[0..n-1]`                          | `True`：重新索引（常用于垂直合并样本表）                      |
| `keys`             |               `None` | 给每个对象加上顶层标签，生成 **分层索引（MultiIndex）**                    | `keys=['A','B']` → 行索引或列索引前端加入 `'A'/'B'`     |
| `levels`           |               `None` | 与 `keys` 配合，为 MultiIndex 指定 levels（通常不用）               | 稀有用，主要在复杂 MultiIndex 构建时指定                   |
| `names`            |               `None` | 给生成的 MultiIndex 指定每一层的名字                               | 例如 `names=['source','row']`                  |
| `verify_integrity` |              `False` | 是否检查合并后索引是否有重复（若 True，发现重复会报错）                         | `True`：安全但代价高（会触发 O(n) 检查）                   |
| `sort`             | `False`（pandas 1.0+） | 是否对非对齐轴进行排序（当 `join='outer'` 并且列/索引不一致时）               | `sort=True` 会按字典序排序列标签；`False` 可保持输入顺序（推荐）   |
| `copy`             |               `True` | 是否复制数据到新对象（可能占用内存）                                     | `False` 可能返回视图（不常用）                          |
| `join_axes`        |           deprecated | 旧版用于强制使用某轴的索引；已弃用，请用 `reindex` 或 `join` 替代             | —                                            |


常见场景与推荐参数（简要列举）:

| 场景                   | 推荐用法                                               |
| -------------------- | -------------------------------------------------- |
| 将多个相同列结构的样本表纵向合并（堆栈） | `pd.concat([df1, df2], axis=0, ignore_index=True)` |
| 将按样本索引的不同特征表横向合并     | `pd.concat([dfA, dfB], axis=1, join='outer')`      |
| 合并不同列但只关心共有列         | `pd.concat([df1, df2], axis=0, join='inner')`      |
| 合并并保留来源标签（来源为组信息）    | `pd.concat(objs, keys=['g1','g2'])`                |
| 合并大量文件并确保无重复索引       | `pd.concat(objs, verify_integrity=True)`（注意性能）     |


#### 4.9.1.1. <a id='toc4_9_1_1_'></a>[DataFrame and DataFrame](#toc0_)

````python
import pandas as pd
import numpy as np


df1 = pd.DataFrame({chr(num): [contens for contens in np.arange(0, 5)] 
                    for num in np.arange(97, 102) })

df2 = pd.DataFrame({chr(num): [contens for contens in np.arange(0, 5)] 
                    for num in np.arange(100, 105) })
df1
````

````python
df2
````

````python
pd.concat(objs=[df1, df2], 
          axis = 0,           # 纵向
          join = 'outer'      # 并集
) # 默认时候的参数
````

````python
pd.concat(objs=[df1, df2], 
          axis = 0,       # 纵向
          join = 'inner'  # 交集
)
````

````python
pd.concat(objs=[df1, df2], 
    axis=1,         # 横向
    join='outer'    # 并集
)
````

````python
pd.concat(objs=[df1, df2], 
    axis = 1,           # 横向
    join = 'inner'      # 交集
)
````

### 4.9.2. <a id='toc4_9_2_'></a>[merge-关键字（依照关键字合并）](#toc0_)


* <span style="color:blue">依照`关键字(on='key')`进行连接；而不是依照整个`轴(axis=0/1)`</span>

* `df1.merge(df2, on=''共有列名', how='inner/outer/left/right')`

* `pd.merge(left=df1, right=df2, on='共有列名', how='inner/outer/left/right')` # 同上

* `pd.merge(left=df1, right=df2, left_on='left_key', right_on='right_key', how='inner')` # left_index=True/right_index=True，更灵活

* merge后的left_on和right_on (或left_index和right_index)`会消失`

| 参数                                     |                              默认 | 含义（逻辑）                                          | 备注/常见取值                                                             |
| -------------------------------------- | ------------------------------: | ----------------------------------------------- | ------------------------------------------------------------------- |

* concat：
concat 是 Pandas 提供的方法，用于沿着`某一轴`（通常是行轴）简单堆叠数据框或系列对象。
缺省情况下，concat 是纵向拼接，即沿着行方向连接数据。
可以通过设置 axis 参数来实现横向拼接（即沿着列方向连接）。
concat 的结果通常包含沿着指定轴的所有数据行或列。
它不涉及业务层面的连接，更多停留在物理融合的层面，没有参与感。

* merge：
merge 用于按照`指定的键（key）`进行连接操作。
可以指定连接方式（inner、outer、left、right）以及连接的键。
结果可能包含来自左侧和右侧数据框的所有行，或者只包含合并列匹配的行（取决于合并方式）。
merge 更深入地通过共同的索引或共同项，将两组数据从业务层面进行拼接。

## 4.11. <a id='toc4_11_'></a>[分组、聚合操作](#toc0_)

📌 核心思想: 把数据按照某种分组方式归类，并对每类数据执行某种运算（总和、平均、统计等）。

例如：

| SampleID | OTU  | Abundance |
| -------- | ---- | --------- |
| S1       | OTU1 | 10        |
| S1       | OTU2 | 0         |
| S2       | OTU1 | 3         |

按 SampleID 聚合求总丰度：

| SampleID | Total   |
| -------- | ------- |
| S1       | 10+0=10 |
| S2       | 3       |


分组：
```python
grouped_df = df.groupby(by=['Group', 'Category'], as_index=False, sort=False, dropna=False, observed=False, group_keys=True, squeeze=False, axis=0, level=None)
```

聚合：
- groupby + agg：用于降维汇总（每组一个或多个统计值）
- groupby + transform：用于逐行广播（保留原始行数但加上组统计/规范化）
- groupby + apply：用于复杂自定义逻辑（最灵活但慢）
- groupby + filter：用于按组保留/删除整组原始行
- 其它（size,count,nunique,cum...）= groupby 的快捷函数

路径：
```diff
df.groupby(keys)
        ↓
     子集组 
        ↓
  决定你想要什么结果？
        ↓
- agg       → 得到每组的统计汇总（小表）
- transform → 得到与原表同长度的结果（添加新列）
- apply     → 复杂但灵活，返回任意形状
- filter    → 删除/保留整个组
```

### 4.11.2. <a id='toc4_11_2_'></a>[聚合](#toc0_)

##### 4.11.2.1.1. <a id='toc4_11_2_1_1_'></a>[aggregate](#toc0_)

这是最常用最强大的形式, `groupby + agg(func=多函数、多列)`:
- func=接受
  - `字符串`（'sum'、'mean'等）
  - `列表`（[np.sum, np.mean]）
  - `函数`（np.sum）, 自定义函数（lambda x: x.max() - x.min()）
  - `字典`（{'col1': 'sum', 'col2': ['mean', 'max']}）
  - `命名聚合`（Named Aggregation）: func= (mean=('col1', 'mean'), sum_col2=('col2', 'sum'))

1. 对单列进行单个聚合:

````python
taxatable.groupby(['Kingdom', 'Genus'])['Domain'].aggregate(func='count')
````

2. 对单列进行多个聚合:

````python
taxatable.groupby(['Kingdom', 'Genus'])['Domain'].aggregate(func=['count', 'sum'])
````

3. 对多列进行多聚合：

````python
taxatable.groupby(['Kingdom', 'Genus']).aggregate(func={'Domain': ['count', 'sum'], 'Phylum': 'max'}) # func接受字典，指定不同列的聚合函数
````

4. 最佳写法：命名聚合（推荐）：

````python
taxatable.groupby(['Kingdom', 'Genus']).aggregate(mean_abund=("Domain", "count"), n=("Kingdom", "count"), ) # test=("Phylum", ["max", 'count'])会报错。
````

5. ✔ agg 的特点：
- 输出 比原表小（降维）
- 输出的行数 = 分组数量
- 输出的列数 = 自己定义或函数数量
- 用于 “分组统计、summary table”

##### 4.11.2.1.2. <a id='toc4_11_2_1_2_'></a>[transform](#toc0_)

✔ transform 的核心作用: 
- 每组返回一个与该组同长度的结果，最后拼回原表，保持原表行数不变；
- **组内计算 + 广播回原表**。


✔ transform vs agg 的本质区别：

| 操作            | 返回大小           | 典型用途          |
| ------------- | -------------- | ------------- |
| **agg**       | 每组 1 个值 → 行数减少 | 统计、summary    |
| **transform** | 每行都有值 → 行数不变   | 归一化、标准化、填充、标记 |


````python
taxatable.groupby(by=['Domain']).agg(func='count').head(10)
````

````python
taxatable.groupby(by=['Domain']).transform(func='count').head(10) # func=lambda x: x.count()
````

##### 4.11.2.1.3. <a id='toc4_11_2_1_3_'></a>[apply](#toc0_)

✔ apply 的作用:
- 对每个组执行函数，函数可以返回：
  - 标量
  - Series
  - DataFrame
  - 多行多列
- pandas 会自动拼接
- 最灵活，但是很慢，输出不可预测(除非设计严谨)，常用于复杂的逻辑：如bootstrap、模型拟合、挑选top-K样本

````python
def top_row(df):
    idx = df['Kingdom'].idxmax()
    return df.loc[idx]

taxatable.groupby(by=['Domain']).apply(func=top_row)
````

##### 4.11.2.1.4. <a id='toc4_11_2_1_4_'></a>[filter](#toc0_)

✔ filter 的作用:
- 据条件决定：
  - 整个组保留
  - 整个组丢弃
- 非常适合微生物组数据中的 prevalence filtering。

````python
taxatable.groupby(by=['Domain']).filter(func=lambda x: (x.count() > 0).sum() > 1).head(10)
````

##### 4.11.2.1.5. <a id='toc4_11_2_1_5_'></a>[四者间的关系（终极表）](#toc0_)

| 操作            | 返回形状    | 是否保留原行数   | 用途               | 例子       |
| ------------- | ------- | --------- | ---------------- | -------- |
| **agg**       | 每组一条或几条 | ❌ 不保留     | 统计 summarization | 平均丰度     |
| **transform** | 与原表同样行数 | ✔ 保留      | 组内归一化、标记         | 相对丰度     |
| **apply**     | 不确定     | 不确定       | 复杂处理             | 挑选 top-N |
| **filter**    | 原表子集    | ❌（只保留部分组） | prevalence 过滤    | 出现频率     |


##### 4.11.2.1.6. <a id='toc4_11_2_1_6_'></a>[agg vs transform](#toc0_)

假设表：

| Class | Score |
| ----- | ----- |
| A     | 60    |
| A     | 80    |
| B     | 70    |
| B     | 90    |
| B     | 70    |

📌 groupby('Class').agg('mean')：
- 分组数量 = 2（A, B） → 输出 2 行
- 这就是 agg → 降维

| Class | mean  |
| ----- | ----- |
| A     | 70    |
| B     | 76.67 |

📌 groupby('Class').transform('mean')：
- 行数 = 原表 5 行完全不变
- 这就是 transform → 不降维

| Class | Score | mean  |
| ----- | ----- | ----- |
| A     | 60    | 70    |
| A     | 80    | 70    |
| B     | 70    | 76.67 |
| B     | 90    | 76.67 |
| B     | 70    | 76.67 |

#### 5.2.3.1. <a id='toc5_2_3_1_'></a>[agg](#toc0_)

````python
grouped = (
    otu_meta
    .groupby(by=['OTU_ID', 'SoilType']) # There is difference in order by=['OTU_ID', 'SoilType'], but difference is excluded when reset_index().
    .agg(mean_abun=("Abundance", "mean"), sd_abun=("Abundance", "std"), n=("Abundance", "count"))
    .reset_index()
)
grouped.head()
````

````python
grouped_pivot_table = grouped.pivot_table(
    index='OTU_ID',
    columns='SoilType',
    values=['mean_abun', 'sd_abun', 'n'],
)
grouped_pivot_table.head()
````
