# 2. <a id='toc2_'></a>[绘图原理](#toc0_)


## 2.1. <a id='toc2_1_'></a>[模块、绘图逻辑](#toc0_)

matplotlib绘图的原理：在画布（figure）上绘制坐标系（axes），其中axes上的各个元素如下：

<div style="display:flex;justify-content:center">

<img src="https://matplotlib.org/stable/_images/anatomy.png" alt="../../_images/anatomy.png" width=500 height=500 />

</div>

**注意区分:**  
  1. `Axes`:整个坐标系；
  2. `Axis`:坐标系上的坐标轴；
  3. `Artist`:在figure上看到的所有元素都是Artist，包括：Figure，Axes，Axis，Text，Line2D，collections，Path对象等。

MATPLOTLIB的模块逻辑：

1. matplotlib绘图逻辑：

```bash
        Scripting Layer（pyplot）
                ↓
        Artist Layer:
                Figure                          # 管理画布、多个Axes、DPI、尺寸
                └── Axes                        # 正真绘图区、管理坐标系、管理所有子Artist
                        ├── Axis (xaxis / yaxis)    # x/y轴、管理tick、管理刻度标签
                        ├── Line2D                  # 具体Artist
                        ├── Text                    # 具体Artist
                        ├── Patch                   # 具体Artist
                        └── Collection              # 具体Artist
                ↓
        Backend Layer:
                ├── GUI backend: TKAgg, Qt5Agg, MacOSX  # 支持交互窗口、鼠标操作、缩放
                ├── Non-GUI backend: Agg, PDF, SVG      # 用于渲染和保存图片
```


2. matplotlib包架构：

```bash
        matplotlib/
        │
        ├── pyplot.py
        ├── figure.py                   # class Figure
        ├── axes/                       # class Axes
        ├── axis.py
        ├── lines.py                    # class Line2D and etc.
        ├── text.py
        ├── patches.py
        ├── collections.py
        ├── backend_bases.py
        ├── backends/
        └── transforms.py               # 坐标变换
```

## 2.2. <a id='toc2_2_'></a>[作图的方式](#toc0_)

快速绘图可以用`plt`方式，而大型项目可以用`面向对象`的方式进行绘图。

### 2.2.1. <a id='toc2_2_1_'></a>[基于面向对象](#toc0_)

实例化 `fig:Figure` 和 `ax:Axes`，并调用对象的绘图方法进行绘制：

````python
x = np.linspace(0, 2, 100)

# Note that even in the OO-style, we use `.pyplot.figure` to create the figure.
# ax.flat属性很有用
fig, ax = plt.subplots(constrained_layout=True, figsize=(3*1.2, 3))                                  # Create a figure and an axes.

ax.plot(x, x, label='linear', linestyle='-', color='red')                   # Plot some data on the axes.
ax.plot(x, x**2, label='quadratic', linestyle='--', color='blue')           # Plot more data on the axes...
ax.plot(x, x**3, label='cubic', linestyle='-.', color='green')              # ... and some more.

ax.set_xlabel('x label')                                                    # Add an x-label to the axes.
ax.set_ylabel('y label')                                                    # Add a y-label to the axes.
ax.set_title("Simple Plot")                                                 # Add a title to the axes.
ax.legend()                                                                 # Add a legend.
````

### 2.2.2. <a id='toc2_2_2_'></a>[基于pyplot接口](#toc0_)

据说是类似于Matlab方式，但是我没有用过，就是不用进行类的实例化，只需要` pyplot (plt)` 调用各种方法即可，其它交给plt自动进行管理：

````python
x = np.linspace(0, 2, 100)

plt.figure(figsize=(3*1.2, 3))
plt.plot(x, x, label='linear')          # Plot some data on the (implicit) axes.
plt.plot(x, x**2, label='quadratic')    # etc.
plt.plot(x, x**3, label='cubic')

plt.xlabel('x label')
plt.ylabel('y label')
plt.title("Simple Plot")
plt.legend()
````

# 3. <a id='toc3_'></a>[布局 (Layout)](#toc0_)



## 3.1. <a id='toc3_1_'></a>[方式一（推荐）：subplot()挨个规划](#toc0_)

```python
fig = plt.figure() -> Figure
ax0 = plt.subplot(nrows, ncols, index) -> Axes
  # projection = '3d' or 'polar'
  # sharex/y = ax1
```

````python
# pyplot方式进行绘图（其实是pyplot和面向对象混用）
plt.figure(constrained_layout=True)             # 创建一个figure画布

ax1 = plt.subplot(331)                          # 创建一个axes，因为subplot返回值就是axes类型，其实也是通过add_subplot()实现的（原函数）
                                                # ax1也可以不用指出，此处为了方便后续共享坐标轴
                                                # row = 3, col = 3, index = 1
                                                # 共享坐标轴：sharex=ax1, sharey=ax1,这是非常有用的
plt.plot(range(0,20))
plt.xlabel("x")
plt.ylabel("y")
plt.title('331')

## 子图
## 共享ax1的X轴
plt.subplot(332, sharex= ax1, sharey= ax1)      # 创建一个axes，并指明与ax1共享坐标轴
plt.scatter(range(0, 10),range(0,10))
plt.xlabel("x")
plt.ylabel("y")
plt.title("332 (share x/y with 321) ")

## 子图
plt.subplot(312)                                # 创建一个axes，不用共享坐标轴
plt.plot(range(0, 10))
plt.xlim([0, 15])                                                                       # xlim,plt.xlim(xmin,xmax)
plt.ylim([0, 15])                                                                       # ylim
plt.grid()                                                                              # grid
plt.axhline(y=3, c='r')                                                                 # hor:水平线
plt.axvline(x=3, c='Blue')                                                              # ver:垂直线
plt.axhspan(ymin=4.5, ymax=9.0, facecolor='green', alpha=0.5)                           # 水平区间
plt.axvspan(xmin=4.5, xmax=9.0, facecolor='gray', alpha=0.6)                            # 垂直区间
plt.annotate(text='End point',xy=(9,9),xytext=(10,9.5),arrowprops={'color':'r'})        # 注释文本
plt.text(x=3,y=3,s='Point haha')                                                        # 简单的文本说明，注释文本
plt.xlabel("xlabel")
plt.ylabel("ylabel")
plt.title("Title is 312")

## 子图
ax4 = plt.subplot(313)                          # 创建一个axes，不用共享坐标轴，混合面向对象编程
ax4.plot(range(0, 10))
ax4.set_xlim([0,15])
ax4.set_ylim([0,15])
ax4.grid(True,axis='x')
ax4.axhline(y=3, c='r')
ax4.axvline(x=3, c='Blue')
ax4.axhspan(ymin=4.5, ymax=9.0, facecolor='green', alpha=0.5)
ax4.axvspan(xmin=4.5, xmax=9.0, facecolor='gray', alpha=0.6)
ax4.annotate(text='End point',xy=(9,9), xytext=(10,9.5), arrowprops={'color':'r'})
ax4.text(x=3,y=3,s='Point haha')
ax4.set_xlabel('x')
ax4.set_ylabel('y')
ax4.set_title('Title is 313')
ax4.spines['right'].set_color('None')                                                   # 取消right
ax4.spines['top'].set_color('None')                                                     # 取消top

plt.suptitle("This is a figure suptitle.")                                                   # figure级别的标题
````

## 3.7. <a id='toc3_7_'></a>[GridSpec (推荐)](#toc0_)

应对复杂布局：
- 不等大小子图
- 合并子图
- 自定义比列
- 复杂论文排版
- `GridSpec只负责切分出网格，利用constrained_layout进行自动排列`

### 3.7.1. <a id='toc3_7_1_'></a>[GridSpec()](#toc0_)

````python
from matplotlib import gridspec


fig = plt.figure(constrained_layout=True, figsize=(3*1.2, 3))

gs = gridspec.GridSpec(
    3, 3, figure=fig, 
    # wspace=0.5, hspace=0.5,                   # 调整子图间距
    left=0.1, right=0.9, top=0.9, bottom=0.1,   # 调整figure边距
    height_ratios=[1, 2, 1],                    # 行高比例，第一行和第三行的高度是第二行的一半
    width_ratios=[1, 2, 1],                     # 列宽比例，第一列和第三列的宽度是第二列的一半 
)

ax1 = fig.add_subplot(gs[0, 0]) 
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, :])
ax4 = fig.add_subplot(gs[2, :])
````

### 4.1.3. <a id='toc4_1_3_'></a>[Figure.colorbar](#toc0_)

Figure.colorbar() 是 Matplotlib 中 Figure 对象的方法，用于在整张图（figure）级别创建 colorbar（颜色条），并把它与某个 ScalarMappable（例如 imshow、contourf、scatter 等返回对象）关联。

该方法在 Figure 对象上添加一个颜色条。与 pyplot.colorbar 不同，它需要明确指定所属的 Figure 对象，适合面向对象的编程风格。

```python
Figure.colorbar(mappable, cax=None, ax=None, use_gridspec=True, **kwargs)
```

重要参数：
| 参数             | 类型             | 说明                                 |
| -------------- | -------------- | ---------------------------------- |
| `mappable`     | ScalarMappable | 需要映射颜色的对象（如 `imshow` 返回值）          |
| `cax`          | Axes           | 指定 colorbar 所在的 Axes               |
| `ax`           | Axes 或 Axes列表  | 指定与哪些 Axes 关联                      |
| `use_gridspec` | bool           | 是否使用 GridSpec 自动布局                 |
| `orientation` | str           | horizontal或vertical                 |
| `fraction` | float           | 控制colorbar宽度比例                 |
| `pad` | float           | colorbar 与图之间的间距                 |
| `shrink` | float           | 缩放 colorbar 长度                |
| `aspect` | float           | 控制 colorbar 的长宽比                |
| `ticks` | list           | 自定义刻度, ticks=[0,0.5,1]               |
| `format` | stf           | 刻度格式, format="%.2f"               |
| `**kwargs`     | dict           | 传递给 `matplotlib.colorbar.Colorbar` |

什么是 mappable？
| 绘图函数           | 返回对象           | 是否可用于 colorbar |
| -------------- | -------------- | -------------- |
| `imshow()`     | AxesImage      | ✔              |
| `pcolormesh()` | QuadMesh       | ✔              |
| `scatter()`    | PathCollection | ✔              |
| `contourf()`   | QuadContourSet | ✔              |

总结：
| 功能          | 说明                               |
| ----------- | -------------------------------- |
| 创建 colorbar | `Figure.colorbar()`              |
| 绑定对象        | `mappable`                       |
| 控制位置        | `ax` / `cax`                     |
| 控制大小        | `fraction`, `shrink`, `aspect`   |
| 控制样式        | `ticks`, `format`, `orientation` |
| 返回对象        | `Colorbar`                       |


每个Axes单独的colorbar：

````python
cmaps = ['RdBu_r', 'viridis']
fig, axs = plt.subplots(2, 2, layout='constrained', figsize=(3*1.2, 3))
for col in range(2):
    for row in range(2):
        ax = axs[row, col]
        pcm = ax.pcolormesh(np.random.random((20, 20)) * (col + 1), cmap=cmaps[col])
        
        fig.colorbar(mappable=pcm, ax=ax, orientation='horizontal', fraction=0.1, pad=0.03, shrink=1)
````

多个Axes共享：

````python
cmaps = ['RdBu_r', 'viridis']
fig, axs = plt.subplots(2, 2, layout='constrained', figsize=(3*1.2, 3))
for col in range(2):
    for row in range(2):
        ax = axs[row, col]
        pcm = ax.pcolormesh(np.random.random((20, 20)) * (col + 1), cmap=cmaps[col])

    fig.colorbar(pcm, ax=axs[:, col], shrink=0.6)
````

````python
x = np.random.rand(30)
y = np.random.rand(30)
group = np.random.choice(['A', 'B'], size=30)

plt.figure(figsize=(3*1.2, 3))
plt.scatter(x[group=='A'], y[group=='A'], c='blue', label='Group A')
plt.scatter(x[group=='B'], y[group=='B'], c='red', label='Group B')

plt.legend()
````

````python
import matplotlib.pyplot as plt
import numpy as np

labels = ['cat', 'cat', 'dog', 'dog', 'bird']
unique_labels = list(set(labels))
label_to_int = {label: i for i, label in enumerate(unique_labels)}
c = [label_to_int[label] for label in labels]

x = np.random.randn(len(labels))
y = np.random.randn(len(labels))

plt.figure(figsize=(3*1.2, 3))

for label in unique_labels:
    idx = [i for i, l in enumerate(labels) if l == label]
    plt.scatter(x[idx], y[idx],
                c=[c[i] for i in idx], cmap='tab10',
                label=label, s=80)

plt.legend(title="Animal")
# plt.colorbar(label='Prediction Score')
plt.show()
````

🔔 关键点：
- c 必须是一维数组或列表，长度 = 数据点数量（即 len(x) == len(y) == len(c)）
- 使用 cmap 指定离散色图（如 'tab10', 'Set1', 'Paired'）
- 类别数 ≤10 用 'tab10'，≤20 用 'tab20'

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap


N = 50  # 类别数量
colors = plt.get_cmap('tab20', N)  # tab20可重复取色
cmap = ListedColormap(colors(np.linspace(0, 1, N)))

x = np.random.randn(N)
y = np.random.randn(N)
c = np.arange(N)

fig, ax = plt.subplots(constrained_layout=True, figsize=(3*1.2, 3))
sc = ax.scatter(x, y, c=c, cmap=cmap, s=60)
fig.colorbar(sc, label='Class index')
````

### 5.4.2. <a id='toc5_4_2_'></a>[linestyle属性](#toc0_)

````python
import matplotlib.pyplot as plt
import numpy as np


# https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html

linestyle_str = [
     ('solid', 'solid'),      # Same as (0, ()) or '-'
     ('dotted', 'dotted'),    # Same as (0, (1, 1)) or ':'
     ('dashed', 'dashed'),    # Same as '--'
     ('dashdot', 'dashdot')]  # Same as '-.'

linestyle_tuple = [
     ('loosely dotted',        (0, (1, 10))),
     ('dotted',                (0, (1, 1))),
     ('densely dotted',        (0, (1, 1))),
     ('long dash with offset', (5, (10, 3))),
     ('loosely dashed',        (0, (5, 10))),
     ('dashed',                (0, (5, 5))),
     ('densely dashed',        (0, (5, 1))),

     ('loosely dashdotted',    (0, (3, 10, 1, 10))),
     ('dashdotted',            (0, (3, 5, 1, 5))),
     ('densely dashdotted',    (0, (3, 1, 1, 1))),

     ('dashdotdotted',         (0, (3, 5, 1, 5, 1, 5))),
     ('loosely dashdotdotted', (0, (3, 10, 1, 10, 1, 10))),
     ('densely dashdotdotted', (0, (3, 1, 1, 1, 1, 1)))]


def plot_linestyles(ax, linestyles, title):
    X, Y = np.linspace(0, 100, 10), np.zeros(10)
    yticklabels = []

    for i, (name, linestyle) in enumerate(linestyles):
        ax.plot(X, Y+i, linestyle=linestyle, linewidth=1.5, color='black')
        yticklabels.append(name)

    ax.set_title(title)
    ax.set(ylim=(-0.5, len(linestyles)-0.5),
           yticks=np.arange(len(linestyles)),
           yticklabels=yticklabels)
    ax.tick_params(left=False, bottom=False, labelbottom=False)
    ax.spines[:].set_visible(False)

    # For each line style, add a text annotation with a small offset from
    # the reference point (0 in Axes coords, y tick value in Data coords).
    for i, (name, linestyle) in enumerate(linestyles):
        ax.annotate(repr(linestyle),
                    xy=(0.0, i), xycoords=ax.get_yaxis_transform(),
                    xytext=(-6, -12), textcoords='offset points',
                    color="blue", fontsize=8, ha="right", family="monospace")


# fig, (ax0, ax1) = plt.subplots(2, 1, figsize=(10, 8), height_ratios=[1, 3])
fig, (ax0, ax1) = plt.subplots(constrained_layout=True, nrows=2, figsize=(3, 7))

plot_linestyles(ax0, linestyle_str[::-1], title='Named linestyles')
plot_linestyles(ax1, linestyle_tuple[::-1], title='Parametrized linestyles')
````

## 5.7. <a id='toc5_7_'></a>[小提琴图（violinplot）](#toc0_)

```python
matplotlib.pyplot.violinplot(dataset, positions=None, vert=True, widths=0.5, showmeans=False, showextrema=True, showmedians=False, quantiles=None, points=100, bw_method=None, *, data=None)[source]

# =============================================================================
```

````python
import matplotlib.pyplot as plt
import numpy as np


def adjacent_values(vals, q1, q3):
    upper_adjacent_value = q3 + (q3 - q1) * 1.5
    upper_adjacent_value = np.clip(upper_adjacent_value, q3, vals[-1])

    lower_adjacent_value = q1 - (q3 - q1) * 1.5
    lower_adjacent_value = np.clip(lower_adjacent_value, vals[0], q1)
    return lower_adjacent_value, upper_adjacent_value


def set_axis_style(ax, labels):
    ax.set_xticks(np.arange(1, len(labels) + 1), labels=labels)
    ax.set_xlim(0.25, len(labels) + 0.75)
    ax.set_xlabel('Sample name')


# create test data
np.random.seed(19680801)
data = [sorted(np.random.normal(0, std, 100)) for std in range(1, 5)]

# Create a figure instance and two subplots, and unpack the output array immediately
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, constrained_layout=True, figsize=(3*1.2*2, 3), sharey=True)

## ax1 for violinplot
ax1.set_title('Default violin plot', loc='left', fontsize='large')
ax1.set_ylabel('Observed values')
ax1.violinplot(data)

## ax2 for customized violinplot
ax2.set_title('Customized violin plot', loc='left', fontsize='large')
parts = ax2.violinplot(
        data, showmeans=False, showmedians=False,
        showextrema=False)

for pc in parts['bodies']:
    pc.set_facecolor('#D43F3A')
    pc.set_edgecolor('black')
    pc.set_alpha(1)

quartile1, medians, quartile3 = np.percentile(data, [25, 50, 75], axis=1)
whiskers = np.array([
    adjacent_values(sorted_array, q1, q3)
    for sorted_array, q1, q3 in zip(data, quartile1, quartile3)])
whiskers_min, whiskers_max = whiskers[:, 0], whiskers[:, 1]

inds = np.arange(1, len(medians) + 1)
ax2.scatter(inds, medians, marker='o', color='white', s=30, zorder=3)
ax2.vlines(inds, quartile1, quartile3, color='k', linestyle='-', lw=5)
ax2.vlines(inds, whiskers_min, whiskers_max, color='k', linestyle='-', lw=1)

# set style for the axes
labels = ['A', 'B', 'C', 'D']
for ax in [ax1, ax2]:
    set_axis_style(ax, labels)
````

````python
%config InlineBackend.figure_format = 'svg'
import matplotlib.pyplot as plt 
from matplotlib.lines import Line2D
from matplotlib.legend import Legend
import numpy as np 


# plot data
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Figue and axes.
fig, ax = plt.subplots(figsize=(3*1.2, 3))

ax.plot(x, y, label='Sine Wave')
ax.plot(x, np.cos(x), label='Cosine Wave')

# Legend 1: default legend
legend_artist_1 = ax.legend(title='Legend 1', loc='upper left', bbox_to_anchor=(1, 1))

# Legend 2: custom legend with Line2D
legend_artist_2 = Legend(
    ax,
    handles=[Line2D([0], [0], color='blue')], 
    labels=['Sine Wave'],
    title='Legend 2', loc='upper left', bbox_to_anchor=(1, 0.7))
ax.add_artist(legend_artist_2)

# Legend 3: custom legend with different markers and colors
legend_artist_3 = Legend(
    ax, 
    handles=[
        Line2D([0], [0], color='r', marker='o', markersize=5, linewidth=2),
        Line2D([0], [0], color='b', marker='s', markersize=5, linewidth=2),
        Line2D([0], [0], color='w', marker='o', markersize=5, markerfacecolor='w', markeredgecolor='r', linewidth=2), # edgecolor为r，其它都用白色隐藏，就是一个圆圈了。
    ],
    labels=['Red Circle', 'Blue Square', 'White Circle with Red Edge'],
    title='Legend 3', loc='upper left', bbox_to_anchor=(1, 0.5))
ax.add_artist(legend_artist_3)

# Legend 4: custom legend with scatter markers
legend_artist_4 = Legend(
    ax, 
    handles=[
        plt.scatter([0, 1, 2], [0, 0, 0], color='g', s=20),
    ],
    labels=['Green Circle'],
    title='Legend 4', loc='upper left', bbox_to_anchor=(1, 0.1))
ax.add_artist(legend_artist_4)

fig.tight_layout()  # 调整布局以适应图例
````

````python
import matplotlib.pyplot as plt
import numpy as np


x = np.linspace(0, 4*np.pi, 200)
y = np.sin(x) * np.exp(-x/10)

plt.figure(figsize=(3*1.2, 3))
plt.plot(x, y, 'b-', linewidth=2)

# 找到最大值点
max_idx = np.argmax(y)
plt.annotate(f'Max: ({x[max_idx]:.1f}, {y[max_idx]:.2f})',
             xy=(x[max_idx], y[max_idx]),
             xytext=(x[max_idx] + 2, y[max_idx] - 0.1),
             arrowprops=dict(
                 arrowstyle='->', 
                 color='red'
             ),
             fontsize=10, color='red')

# 找到最小值点
min_idx = np.argmin(y)
plt.annotate(f'Min: ({x[min_idx]:.1f}, {y[min_idx]:.2f})',
             xy=(x[min_idx], y[min_idx]),
             xytext=(x[min_idx] + 2, y[min_idx] + 0.2),
             arrowprops=dict(
                 arrowstyle='->', 
                 color='green'
            ),
             fontsize=10, color='green')

plt.title("Annotated Extremes")
plt.grid(axis='y', alpha=0.3)
plt.show()
````

`locator` (刻度位置的计算， `定位器`):
- ax.xaxis.set_major_locator() # 主刻度线 数量和排布
- ax.xaxis.set_minor_locator() # 次刻度线 数量和排布
- ax.yaxis.set_major_locator()
- ax.yaxis.set_minor_locator()
- 常用类型：
  - NullLocator()：隐藏刻度（无刻度线）。
  - MaxNLocator(n)：限制刻度最大数量为 n 个，自动调整间隔。
  - MultipleLocator(base)：按固定间隔 base 生成刻度（如 base=0.5 每0.5一个刻度）。
  - LogLocator()：对数坐标下的刻度定位（默认用于 ax.set_xscale('log')）。

`formatter` (刻度标签显示格式， `格式化器`):
- ax.xaxis.set_major_formatter() # 主刻度线上 标签
- ax.xaxis.set_minor_formatter() # 次刻度线上 标签
- ax.yaxis.set_major_formatter()
- ax.yaxis.set_minor_formatter()
- 常用类型：
  - NullFormatter()：隐藏刻度标签（显示刻度线但不显示文字）。
  - FuncFormatter(func)：通过自定义函数 func(value, tick_number) 动态生成标签。
  - DateFormatter(format_str)：日期格式标签（如 '%Y-%m-%d'）。
  - LogFormatter()：对数坐标下的默认标签格式化器。

## 6.4. <a id='toc6_4_'></a>[patheffects](#toc0_)

patheffects 模块用于为任何 Artist（如线条、文本、Patch 对象等）添加渲染效果，例如描边、阴影、下划线、渐变等。它通过对图形的绘制路径进行后处理来实现特效。

- matplotlib.patheffects 常用类 / 函数:

    | 类 / 函数               | 作用                 | 常见参数                                | 示例                                            |
    | -------------------- | ------------------ | ----------------------------------- | --------------------------------------------- |
    | `AbstractPathEffect` | 所有路径效果的基类          | 无                                   | 一般不直接使用                                       |
    | `Normal`             | 默认绘制效果（不做额外处理）     | 无                                   | `path_effects.Normal()`                       |
    | `Stroke`             | 为对象添加描边（outline）   | `linewidth`, `foreground`           | `Stroke(linewidth=3, foreground="black")`     |
    | `SimpleLineShadow`   | 为线条添加简单阴影          | `offset`, `shadow_color`            | `SimpleLineShadow()`                          |
    | `SimplePatchShadow`  | 为 Patch 图形添加阴影     | `offset`, `shadow_rgbFace`, `alpha` | `SimplePatchShadow()`                         |
    | `PathPatchEffect`    | 对 PathPatch 使用路径效果 | `offset`, `facecolor`               | `PathPatchEffect()`                           |
    | `TickedStroke`       | 绘制带刻度的描边           | `angle`, `spacing`, `length`        | `TickedStroke()`                              |
    | `withStroke`         | 简化版描边效果            | `linewidth`, `foreground`           | `withStroke(linewidth=3, foreground="white")` |

- 常见参数说明:

    | 参数             | 作用    | 示例                   |
    | -------------- | ----- | -------------------- |
    | `linewidth`    | 描边宽度  | `linewidth=3`        |
    | `foreground`   | 描边颜色  | `foreground="black"` |
    | `offset`       | 阴影偏移量 | `offset=(2,-2)`      |
    | `shadow_color` | 阴影颜色  | `"gray"`             |
    | `alpha`        | 透明度   | `alpha=0.5`          |
    | `spacing`      | 刻度间距  | `spacing=5`          |
    | `angle`        | 刻度角度  | `angle=45`           |


````python
from matplotlib import patheffects


fig, ax = plt.subplots(figsize=(3*1.2, 3))
text = ax.text(0.5, 0.5, 'Hello World', ha='center', va='center', fontsize=30, color='white')
# 添加路径效果：黑色描边 + 阴影
text.set_path_effects([
    patheffects.Stroke(linewidth=3, foreground='black'),  # 描边
    patheffects.Normal(),                                 # 原始文本（白色）
    patheffects.SimplePatchShadow(offset=(2, -2), shadow_rgbFace='gray')  # 阴影
])

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
````

## 8.1. <a id='toc8_1_'></a>[颜色基本知识](#toc0_)

Hues: 波长 (RGB极其组合)，0-360° 或 [0, 1]，Red: 0°, Green: 120°, Blue: 240°；

Saturation: 纯色与灰色的比值，[0, 1] 或 [0, 100%]；

Brightness: 入射或反射光的量，[0, 1] 或 [0, 100%]。

````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.colors as mcolors
import numpy as np
import colorsys


def show_color(n_hue: int = 12, type: str = 'brightness', n_value: int = 6, gap: int = 4, block_size: int = 60):
    # ==================== 参数设置 ====================
    # n_hue = 24          # 从左到右的色相数量（不同颜色）
    # n_value = 12         # 从上到下的亮度级别（同一颜色的亮度变化）
    # gap = 4             # 色块之间的间隙（像素）
    # block_size = 60     # 每个色块的边长（像素）

    # 计算总图像尺寸（像素）
    width_px = n_hue * block_size + (n_hue - 1) * gap
    height_px = n_value * block_size + (n_value - 1) * gap

    # 转换为英寸（假设 DPI=100，可根据需要调整）
    dpi = 100
    fig_width = width_px / dpi
    fig_height = height_px / dpi

    fig, ax = plt.subplots(figsize=(fig_width, fig_height), layout='constrained', dpi=dpi)
    ax.set_xlim(0, width_px)
    ax.set_ylim(0, height_px)
    ax.set_aspect('equal')

    # ==================== 生成颜色矩阵 ====================
    for i in range(n_hue):          # i 控制列（不同色相）
        hue = i / n_hue             # 色相在 0~1 之间均匀分布
        for j in range(n_value):    # j 控制行（同一颜色的亮度变化）
            # 亮度从上到下递减（j=0 最亮，j=n_value-1 最暗）
            value = 1.0 - j / (n_value - 1)   # 从 1.0 到 0.0
            # 使用 HLS 颜色空间：色相 hue，饱和度 1.0，亮度 lightness
            if type == 'brightness':
                saturation = 1.0
                lightness = value
            elif type == 'saturation':
                lightness = 0.5
                saturation = value
            r, g, b = colorsys.hls_to_rgb(hue, lightness, saturation)
            color = (r, g, b)
            
            # 计算当前色块在图像中的坐标（左下角原点，y 轴从下往上）
            x = i * (block_size + gap)
            y = j * (block_size + gap)
            rect = patches.Rectangle((x, y), block_size, block_size,
                                    facecolor=color, edgecolor='white', linewidth=0)
            ax.add_patch(rect)
            
            # 可选：在色块中央标注亮度值（便于观察）
            # if j % 2 == 0:   # 避免太拥挤
                # ax.text(x + block_size/2, y + block_size/2, f'{lightness:.1f}',
                #         ha='center', va='center', fontsize=8, color='black' if lightness>0.5 else 'white')

    # ==================== 添加坐标标签 ====================
    # X 轴标签：不同色相
    xticks = [i * (block_size + gap) + block_size/2 for i in range(n_hue)]
    xlabels = [f'Hue\n{i/n_hue:.2f}' for i in range(n_hue)]
    ax.set_xticks(xticks)
    ax.set_xticklabels(xlabels, fontsize=8)

    # Y 轴标签：不同亮度
    yticks = [j * (block_size + gap) + block_size/2 for j in range(n_value)]
    ylabels = [f'{type.capitalize()}\n{1.0 - j/(n_value-1):.2f}' for j in range(n_value)]
    ax.set_yticks(yticks)
    ax.set_yticklabels(ylabels, fontsize=8)

    ax.set_xlabel('Hue (different colors from left to right)', fontsize=10)
    ax.set_ylabel(f'{type.capitalize()}', fontsize=10)
    ax.set_title('Color Matrix: Varying Hue (→) and ' + f'{type.capitalize()}' + ' (↓)', fontsize=12)

    # 隐藏坐标轴边框
    for spine in ax.spines.values():
        spine.set_visible(False)
````

## 8.4. <a id='toc8_4_'></a>[colormap-3分类](#toc0_)


- `plt.cm.XXX`：直接调用 colormap 对象，如 plt.cm.viridis
- 定性颜色映射表：用于区分不同的类别
- 渐变颜色映射表：用于表示数值的大小
- 灰度颜色映射表：用于表示数值的大小


🔢 Matplotlib 官方将 colormaps 分为几类：

| 类别                    | 用途               | 示例                                                           |
| --------------------- | ---------------- | ------------------------------------------------------------ |

### 10.1.1. <a id='toc10_1_1_'></a>[配色逻辑](#toc0_)

我们在计算机中通常用 RGB（红绿蓝）定义一种颜色，但人眼/视觉感知不是按 RGB 强度区分,更符合人类感知的方式是把颜色拆为:
- 色相 (hue)—— 决定 “颜色名字” (“red”, “blue”…)；
- 饱和度/色彩度 (saturation / chroma)—— 决定颜色的鲜艳程度；
- 明度／亮度 (luminance / lightness) 三个维度—— 决定颜色是明亮还是暗。

>原则 A：当你需要区分类别 (categorical) — 应该主要用 hue (色相) 的差异。   
>原则 B：当你需要表达数值 (numeric) 的大小或强度时 — 应该用亮度 / 明度 (luminance) 的变化 (从浅到深)，而不是 hue 的随意变化。

颜色调色板的三大核心类型（正式分类）:

| 类型                            | 适用数据             | 定义                       | 典型用途                  | Seaborn 示例                                    |
| ----------------------------- | ---------------- | ------------------------ | --------------------- | --------------------------------------------- |
| **1. Qualitative（定性 / 离散分类）** | 离散类别（无顺序）        | 颜色彼此无阶梯关系，用于区分类别         | 分类变量（如 Species、Group） | `sns.color_palette("Set2")`、`"tab10"`、`"hls"`、**`"husl"`** |
| **2. Sequential（定量 / 连续数值）**  | 连续递增数据（有顺序）      | 单方向递增的亮度/饱和度，用于大小表示      | 热力图、连续变量              | **`"Blues"`**、`"Greens"`、`"viridis"`              |
| **3. Diverging（发散型）**         | 以一个中心点对称，两端发散的数据 | 中间有中性点（零点），两侧表示 + / - 差异 | Z-score 热图、基因表达差值     | `"RdBu"`、`"PiYG"`、**`"coolwarm"`**                |

````python
def show_contexts():
    contexts = ['paper', 'notebook', 'talk', 'poster']
    fig, axes = plt.subplots(1, len(contexts), figsize=(12, 3))

    for ax, context in zip(axes, contexts):
        with sns.plotting_context(context):                     # 设置不同的context
            sns.barplot(x=['A', 'B', 'C'], y=[1, 3, 2], ax=ax)
            ax.set_title(f'Context: {context}')

    plt.tight_layout()


# Show the different contexts
show_contexts()
````

## 10.3. <a id='toc10_3_'></a>[图表类型](#toc0_)

- Figure-level functions: 
  - catplot(), displot(), pairplot(), relplot(), lmplot()
  - jointplot(), pairgrid(), facetgrid()
  - 有col, col_wrap, row参数，用于创建子图网格,适合多变量数据可视化,legend在图外
  - 调整图表大小：height, aspect (宽高比)

- Axes-level functions: 
  - scatterplot(), lineplot(), histplot(), kdeplot(), boxplot(), violinplot(), heatmap()
  - 没有col, col_wrap, row参数，适合单变量或双变量数据可视化,legend在图内
  - 调整图表大小：plt.figure(figsize=(width, height))

- Figure and Axes-level functions 多变量可视化：
  - hue参数：按类别变量着色
  - hue_order参数：指定类别顺序
  - size参数：按数值变量调整点大小
  - style参数：按类别变量调整点形状 
  - linestyle参数：按类别变量调整线型
  - col, col_wrap, row, row_wrap参数：创建子图网格

# 11. <a id='toc11_'></a>[统计分析 (statannotations)](#toc0_)

````python
from statannotations.Annotator import Annotator


def show_statannotations(ax: plt.Axes, pairs: list, data=None, x=None, y=None):
    annotator = Annotator(
        ax=ax,
        pairs=pairs,
        data=data,
        x=x,
        y=y,
        # order=order,
    )
    annotator.configure(
        test='t-test_ind', 
        text_format='full', 
        loc='inside', 
        fontsize=10, 
        line_width=0.5,
        color= 'red'
    )
    annotator.apply_and_annotate()


# Demo data
data_df: pd.DataFrame = pd.DataFrame({
    'x': ['A', 'B', 'C'],
    'y': [1, 3, 2]
})

fig, axs = plt.subplots(1, 2, figsize=(3*1.2*2, 3), constrained_layout=True)

# 使用统计配置statannotations
params = {
    'data': data_df,
    'x': 'x',
    'y': 'y',
}
sns.barplot(ax=axs[0], **params)
show_statannotations(ax=axs[0], pairs=[('A', 'B'), ('B', 'C'), ('A', 'C')], **params)

# 直接使用p-values来配置statannotations
sns.barplot(ax=axs[1], **params)
show_statannotations(ax=axs[1], pairs=[('A', 'B'), ('B', 'C'), ('A', 'C')], **params)
````
