## <a id='toc4_15_'></a>[Menu](#toc0_)

通过使用菜单控件（Menu）可以充分地节省有限的窗口区域，让我们的界面更加简洁优雅，避免臃肿、混乱。

Tkinter Menu 控件提供了三种类型的菜单，分别是：topleve（主目录菜单）、pull-down（下拉式菜单）、pop-up（弹出式菜单，或称快捷式菜单）。

```python
import tkinter as tk

root = tk.Tk()

# <a id='toc5_'></a>[1️⃣ 创建主菜单](#toc0_)
menubar = tk.Menu(root)

# <a id='toc6_'></a>[2️⃣ 创建子菜单](#toc0_)
file_menu = tk.Menu(menubar, tearoff=0)

# <a id='toc7_'></a>[3️⃣ 添加菜单项](#toc0_)
file_menu.add_command(label="Open")
file_menu.add_command(label="Save")
file_menu.add_separator()
file_menu.add_command(label="Exit", command=root.quit)

# <a id='toc8_'></a>[4️⃣ 挂到主菜单](#toc0_)
menubar.add_cascade(label="File", menu=file_menu)

# <a id='toc9_'></a>[5️⃣ 绑定到窗口](#toc0_)
root.config(menu=menubar)

root.mainloop()
```

## <a id='toc6_1_'></a>[pack](#toc0_)

pack() 是一种较为简单的布局方法，在不使用任何参数的情况下，它会将控件以添加时的先后顺序，自上而下，一行一行的进行排列，并且默认居中显示。

```python
lb.pack(side='top/left/bottom/right', fill='x/y/both', expand=True/False, anchor='')
# <a id='toc10_'></a>[Args:](#toc0_)
# <a id='toc11_'></a>[    side: 从什么方向排列；](#toc0_)
# <a id='toc12_'></a>[    anchor: 向什么方向对齐；](#toc0_)
# <a id='toc13_'></a>[    fill: 向什么方向填充；](#toc0_)
# <a id='toc14_'></a>[    expand: 是否扩展到剩余的所有空间。](#toc0_)

# <a id='toc15_'></a>[anchor有 8 个可选：](#toc0_)
wn------n-----en
|              |
w     center   e
|              |
ws------s-----es
```

## <a id='toc6_2_'></a>[grid (推荐)](#toc0_)

grid() 函数是一种基于网格式的布局管理方法，相当于把窗口看成了一张由行和列组成的表格。当使用该 grid 函数进行布局的时，表格内的每个单元格都可以放置一个控件。，从而实现对界面的布局管理。

```python
lb.rowconfigure(0, weight=0) # 0: 不可调, >0 表示相对重要性比例
lb.columnconfigure(0, weidht=0)

lb.grid(row, column, rowspan, columnspan, sticky='ewns')

# <a id='toc16_'></a>[sticky就是既能对齐也能扩展。](#toc0_)
# <a id='toc17_'></a>[sticky就 4 个可选：](#toc0_)
------n-----
|           |
w           e
|           |
------s-----
```

## <a id='toc6_3_'></a>[place](#toc0_)

place() 方法进行布局管理要更加精细化，通过 place() 布局管理器可以直接指定控件在窗体内的绝对位置，或者相对于其他控件定位的相对位置。其原理从左上角 (x=0, y=0)，x向右递增，y向下递增。

```python
# <a id='toc18_'></a>[绝对位置](#toc0_)
lb.place(x=200, y=100, width=10, height=10)

# <a id='toc19_'></a>[相对位置 (相较于父类)](#toc0_)
lb.place(relx=0.1, rely=0.1, relwidth=0.8, relheight=0.8)

```

## <a id='toc6_4_'></a>[布局控件](#toc0_)

### <a id='toc6_4_1_'></a>[Frame](#toc0_)

Frame 本质上也是一个矩形窗体，同其他控件一样也需要位于主窗口内。我们可以在主窗口内放置多个 Frame 控件，并且每个 Frame 中还可以嵌套一个或者多个Frame，从而将主窗口界面划分成多个区域。

### <a id='toc6_4_2_'></a>[LabelFrame](#toc0_)

LabelFrame 控件也是一种容器类型的控件，它属于是 Frame 控件的变体，因此它们的属性选项大体相同。

在默认情况下，LabelFrame 会在其包裹的子组件周围绘制一个边框和一个标题。当我们想要将一些相关的组件分为一组时，就可以使用 LabelFrame 组件，比如把一系列 Radiobutton（单选按钮）放在一起来供用户选择。

### <a id='toc6_4_3_'></a>[PanedWindow](#toc0_)

PanedWindow（窗格界面）也可以理解成一个特殊的 Frame 控件，它是 Tkinter 8.4 版本后新增的空间管理组件，其主要目的是为其他组件提供一个容器或者框架，从而实现以分块的形式对图形界面进行布局。

与 Frame 控件不同， PanedWindow 允许用户自主地调整界面划分以及每块区域的大小。因此，当您需要让用户自己调节每块区域的大小时，就可以采用 PanedWindow 作为组件载体来进行界面的布局。

不仅如此 PanedWindow 组件还提供了“手柄” 功能（设置参数 showhandle=True 来启用），通过拖动“手柄”图标也可以改变每块区域的大小。

### <a id='toc6_4_4_'></a>[Toplevel](#toc0_)

Topleve 是一个顶级窗口控件（也被称为“子窗体”控件），同样类似于 Frame 控件， 不过该控件会脱离根窗口另行创建一个独立窗口，因此它的存在形式不同于上述其他容器。

Toplevel 控件同样隶属于主窗口的子组件，只是存在形式特殊而已。Toplevel 是主窗口之外的弹出框窗口（通过事件来触发执行），在这个窗口内也可以包含其他组件。但需要注意，顶级窗口并不是必须位于窗口的顶部位置，之所以称其为顶级窗口，是因为相对于主窗口而言，Topleve 位于主窗口的上一层。

## <a id='toc7_1_'></a>[切换页面](#toc0_)

```python
import tkinter as tk

root = tk.Tk()
root.geometry("500x400")

# <a id='toc20_'></a>[=========================](#toc0_)
# <a id='toc21_'></a>[菜单](#toc0_)
# <a id='toc22_'></a>[=========================](#toc0_)
menubar = tk.Menu(root)
root.config(menu=menubar)

# <a id='toc23_'></a>[=========================](#toc0_)
# <a id='toc24_'></a>[内容区](#toc0_)
# <a id='toc25_'></a>[=========================](#toc0_)
main_frame = tk.Frame(root)
main_frame.pack(fill="both", expand=True)

# <a id='toc26_'></a>[页面1](#toc0_)
page_home = tk.Frame(main_frame)
tk.Label(page_home, text="首页").pack()

# <a id='toc27_'></a>[页面2（Usearch）](#toc0_)
page_usearch = tk.Frame(main_frame)
tk.Label(page_usearch, text="Usearch 页面").pack()

# <a id='toc28_'></a>[堆叠](#toc0_)
for page in (page_home, page_usearch):
    page.place(relx=0, rely=0, relwidth=1, relheight=1)

# <a id='toc29_'></a>[切换函数](#toc0_)
def show_page(page):
    page.tkraise()

# <a id='toc30_'></a>[菜单绑定](#toc0_)
menu_usearch = tk.Menu(menubar, tearoff=0)
menu_usearch.add_command(label="打开Usearch", command=lambda: show_page(page_usearch))
menubar.add_cascade(label="Usearch", menu=menu_usearch)

# <a id='toc31_'></a>[默认页面](#toc0_)
show_page(page_home)

# <a id='toc32_'></a>[=========================](#toc0_)
# <a id='toc33_'></a>[状态栏](#toc0_)
# <a id='toc34_'></a>[=========================](#toc0_)
status = tk.Label(root, text="Ready", bd=1, relief="sunken", anchor="w")
status.pack(side="bottom", fill="x")

root.mainloop()
```
