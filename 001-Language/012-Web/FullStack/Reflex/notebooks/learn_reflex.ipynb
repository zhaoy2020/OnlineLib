#### 3.1.2.3. <a id='toc3_1_2_3_'></a>[聚合](#toc0_)

2. 聚合分组等操作 (对比pandas)：

| pandas        | 数据库 / ORM        |
| ------------- | ---------------- |
| `groupby`     | `group_by()`     |
| `mean()`      | `func.avg()`     |
| `sum()`       | `func.sum()`     |
| `count()`     | `func.count()`   |
| `max()`       | `func.max()`     |
| `min()`       | `func.min()`     |
| `filter`      | `WHERE / HAVING` |
| `sort_values` | `order_by()`     |


````python
# from sqlalchemy import func
from sqlmodel import select, func


class State(rx.State):
    '''Operation on aggregated data.'''

    stats: list[dict] = []

    def load_stats(self):
        '''Loading aggregated statistics from the database.'''
        
        with rx.session() as session:
            rows = (
                session.exec(select(User.gender, func.count(User.id)).group_by(User.gender))
                .all()
            )
        
        # Convert the result to a list of dictionaries, for easier use in the frontend.
        self.stats = [
            {"gender": gender, "count": count} for gender, count in rows
        ]
````

- 数据库与pandas：
  - SQLAlchemy = ORM 核心
  - SQLModel 封装 SQLAlchemy + Pydantic
  - pandas 全内存，适合分析和复杂计算
- 基本概念对比：
    | 操作类别          | SQLModel                                                | SQLAlchemy ORM                            | pandas                                           |
    | ------------- | ------------------------------------------------------- | ----------------------------------------- | ------------------------------------------------ |
    | **建表 / 定义模型** | `class User(SQLModel, table=True)`                      | `class User(Base): ...` + `Column(...)`   | `pd.DataFrame(...)`                              |
    | **创建表**       | `SQLModel.metadata.create_all(engine)`                  | `Base.metadata.create_all(engine)`        | ❌ pandas 不能建数据库表                                 |
    | **插入数据**      | `session.add(user); session.commit()`                   | `session.add(user); session.commit()`     | `df.loc[len(df)] = [...]`                        |
    | **按主键查询**     | `session.get(User, 1)`                                  | `session.get(User, 1)`                    | `df[df["id"]==1]`                                |
    | **条件查询**      | `select(User).where(User.age>18)`                       | `session.query(User).filter(User.age>18)` | `df[df["age"]>18]`                               |
    | **多条件查询**     | `select(User).where((User.age>18) & (User.city=="北京"))` | `filter(User.age>18, User.city=="北京")`    | `df[(df["age"]>18) & (df["city"]=="北京")]`        |
    | **聚合函数**      | `func.count() / func.sum() / func.avg()`                | `func.count() / func.sum() / func.avg()`  | `df.groupby("col")["col2"].count()/sum()/mean()` |
    | **分组**        | `select(...).group_by(User.city)`                       | `query(...).group_by(User.city)`          | `df.groupby("city")`                             |
    | **排序**        | `.order_by(User.age.desc())`                            | `.order_by(User.age.desc())`              | `.sort_values("age", ascending=False)`           |
    | **限制条数**      | `.limit(10)`                                            | `.limit(10)`                              | `.head(10)`                                      |
    | **多表关联**      | `.join(OtherTable)`                                     | `.join(OtherTable)`                       | `pd.merge(df1, df2, on="key")`                   |
    | **更新**        | `user.age = 20; session.add(user); session.commit()`    | 同 SQLModel                                | `df.loc[df["id"]==1, "age"]=20`                  |
    | **删除**        | `session.delete(user); session.commit()`                | 同 SQLModel                                | `df = df[df["id"]!=1]`                           |
    | **自定义聚合**     | `func.custom(...)` 或 Python 后处理                         | 同 SQLModel                                | `.agg(lambda x: ...)`                            |
    | **适用场景**      | Web API + 数据库，类型安全，自动验证                                 | 纯 ORM 数据库操作，灵活                            | 内存内数据分析，统计，复杂计算                                  |
    | **性能**        | 数据库大数据处理快                                               | 数据库大数据处理快                                 | 大数据可能慢，内存占用高                                     |

- 所有操作对比：
    | 操作                 | SQLModel                         | SQLAlchemy                    | pandas                             |
    | ------------------ | -------------------------------- | ----------------------------- | ---------------------------------- |
    | CRUD               | add, get, delete, commit         | add, query, delete, commit    | loc, append, drop                  |
    | Query              | select(...).where(...)           | query(...).filter(...)        | df[df["col"]>x]                    |
    | Filter             | Python boolean + SQLAlchemy func | filter / filter_by            | Boolean indexing                   |
    | Group/Aggregate    | group_by + func.count/sum/avg    | group_by + func.count/sum/avg | groupby + count/sum/mean           |
    | Sort               | order_by                         | order_by                      | sort_values                        |
    | Limit              | limit()                          | limit()                       | head()                             |
    | Join               | join(...)                        | join(...)                     | merge(...)                         |
    | Update             | get → modify → commit            | get → modify → commit         | loc[...] = ...                     |
    | Delete             | delete → commit                  | delete → commit               | df = df[df...]                     |
    | Custom Aggregation | func + Python                    | func + Python                 | agg(lambda x: ...)                 |
    | Use Case           | Web API + DB, type safe          | pure ORM DB operations        | Data analysis, memory resident     |
    | Performance        | Fast on DB                       | Fast on DB                    | Memory heavy, slower on large data |
- 性能对比：
    | 维度         | 数据库                                | pandas                      |
    | ---------- | ---------------------------------- | --------------------------- |
    | 性能         | 优秀，能处理`百万级`以上数据                      | `内存`占用大，百万级数据会慢               |
    | 灵活度        | 聚合函数`固定`，需要 ORM 或原生 SQL 组合           | `高度灵活`，自定义函数随便写               |
    | 数据流        | 通常 -> 数据库聚合 -> Python/pandas -> UI | 数据全部拉到内存 -> pandas 处理 -> UI |
    | 实时性        | 数据库直接查询`最新数据`                        | `必须先 load 数据到 pandas`         |
    | 多表/复杂 join | `天生支持  `                             | pandas 可以 merge，但逻辑更`复杂、慢`    |

### 3.2.1. <a id='toc3_2_1_'></a>[base](#toc0_)

````python
class CounterState(rx.State):
    '''Auto-refreshing state to hold the counter value.'''

    users: list[User] = [
        User(name="Alice", email="alice@example.com", gender="female"),
        User(name="Bob", email="bob@example.com", gender="male"),
        User(name="Charlie", email="charlie@example.com", gender="male"),
    ]
    count: int = 0

    @rx.event
    def increment(self):
        '''事件处理函数'''
        self.count += 1

    @rx.event
    def decrement(self):
        '''事件处理函数'''
        self.count -= 1

    @rx.var
    def name(self) -> str:
        '''计算属性'''
        return f"Count is {self.count}"
````

## 3.3. <a id='toc3_3_'></a>[认证功能后端](#toc0_)

- Reflex 的认证核心思路：
  - 基于 State + Session + 装饰器控制页面访问
  - 关键：
    - 用户是否登录 = State 里有没有 user / token
    - 登录 / 登出 = 改 State
    - 保护页面 = 检查 State，没有就重定向
  - 👉 Reflex 没有强制你用某种认证方式：
    - 本地用户名 / 密码
    - JWT
    - OAuth（Google / GitHub）
    - 后端 API

````python
import reflex as rx


# Authentication State
class AuthState(rx.State):
    '''State to manage user authentication.'''

    is_authenticated: bool = False
    current_user: str | None = None

    @rx.event
    def login(self, username: str, password: str):
        '''Simulate a login event.'''

        # In a real application, you would verify the username and password.
        if username == "admin" and password == "password":
            self.is_authenticated = True
            self.current_user = username

    @rx.event
    def logout(self):
        '''Simulate a logout event.'''

        self.is_authenticated = False
        self.current_user = None


# Pages
def require_login():
    if not AuthState.is_authenticated:
        return rx.redirect("/login")
    
@rx.page(on_load=require_login)
def Dashboard():
    return rx.box(
        rx.text(f"Welcome, {AuthState.current_user}!"),
        rx.button("Logout", on_click=AuthState.logout),
    )


# Navbar
def Navbar():
    if AuthState.is_authenticated:
        return rx.box(
            rx.text(f"Logged in as {AuthState.current_user}"),
            rx.button("Logout", on_click=AuthState.logout),
        )
    else:
        return rx.box(
            rx.button("Login", on_click=lambda: rx.navigate("/login")),
        )
````

#### 4.1.3.6. <a id='toc4_1_3_6_'></a>[rx.flex](#toc0_)

The Flex component is used to make flexbox layouts. It makes it simple to arrange child components in horizontal or vertical directions, apply wrapping, justify and align content, and automatically size components based on available space, making it ideal for building responsive layouts.

By default, children are arranged horizontally (direction="row") without wrapping.

[https://reflex.dev/docs/library/layout/flex/](https://reflex.dev/docs/library/layout/flex/)

````python
rx.flex(
    rx.card("Card 1"),
    rx.card("Card 2"),
    rx.card("Card 3"),
    rx.card("Card 4"),
    rx.card("Card 5"),
    
    spacing="2",
    width="100%",
    direction="row",
)
````

````python
rx.flex(
    rx.foreach(
        rx.Var.range(10),
        lambda i: rx.card(f"Card {i + 1}", width="16%"),
    ),
    spacing="2",
    flex_wrap="wrap",
    width="100%",
)
````

#### 4.1.3.7. <a id='toc4_1_3_7_'></a>[stack](#toc0_)

Stack is a layout component used to group elements together and apply a space between them.

vstack is used to stack elements in the vertical direction.

hstack is used to stack elements in the horizontal direction.

stack is used to stack elements in the vertical or horizontal direction.

These components are based on the flex component and therefore inherit all of its props.

The stack component can be used with the flex_direction prop to set to either row or column to set the direction.

[https://reflex.dev/docs/library/layout/stack/#stack](https://reflex.dev/docs/library/layout/stack/#stack)

````python
rx.vstack(
    rx.heading('heading'),
    rx.text('text'),
    rx.button('button'),
    rx.progress(value=70),
    
    position='fixed',
    top='0',
    right='0',
    bottom='1em',
    left='0',
    padding='1em',
    margin='0px',
)
````

### 4.2.3. <a id='toc4_2_3_'></a>[Matplotlib](#toc0_)

[https://reflex.dev/docs/library/graphing/other-charts/pyplot/](https://reflex.dev/docs/library/graphing/other-charts/pyplot/)

````python
import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 300

from reflex_pyplot import pyplot # Show in Reflex app.


def create_plot(format: str = 'svg'):
    fig, ax = plt.subplots()
    ax.plot([1, 2, 3], [4, 5, 6])
    ax.set_title("Simple Plot")

    plt.close(fig)  # Not closing the figure could cause memory issues.

    if format == 'svg':
        import io
        import base64 
        buf = io.BytesIO()
        fig.savefig(buf, format='svg')
        img_bytes = buf.getvalue()
        img_base64 = base64.b64encode(img_bytes).decode()
        img_src = f"data:image/svg+xml;base64,{img_base64}"

        return img_src
    else:
        return fig


# show plot in Reflex app
def show_matplotlib(format: str = 'svg'):
    if format == 'svg':
        # show with base64.
        return rx.image(src=create_plot(format=format), alt="Matplotlib SVG Plot")
    else:
        # show with pyplot fig.
        return pyplot(data=create_plot(format=format))
````

### 4.2.4. <a id='toc4_2_4_'></a>[Plotly](#toc0_)

[https://reflex.dev/docs/library/graphing/other-charts/plotly/](https://reflex.dev/docs/library/graphing/other-charts/plotly/)
```bash
uv pip install reflex-pyplot
```

- 需要安装reflex-pyplot包，使用rx.pyplot(data=fig)在reflex中显示图表；
- plotly的expression和seaborn的语法类似；

比较seaborn和plotly的expression：
| seaborn           | plotly（对应）                 | 你要换的“脑回路”              |
| ----------------- | -------------------------- | ---------------------- |
| `sns.lineplot`    | `px.line`                  | 描述关系 → 声明图表            |
| `sns.scatterplot` | `px.scatter`               | 同上                     |
| `sns.barplot`     | `px.bar`                   | seaborn 会算，plotly 要你明确 |
| `data=`           | 第一个参数                      | 少写一个关键词                |
| `hue=`            | `color=`                   | 名字变了                   |
| `style=`          | `line_dash=` / `symbol=`   | 显式控制                   |
| `size=`           | `size=`                    | 基本一致                   |
| `col` / `row`     | `facet_col` / `facet_row`  | 名字变长                   |
| `palette=`        | `color_discrete_sequence=` | 更底层                    |

| seaborn 参数     | plotly.express 参数                    | 含义是否一致  |
| -------------- | ------------------------------------ | ------- |
| `data=df`      | `df`（第一个参数）                          | ✅       |
| `x="x"`        | `x="x"`                              | ✅       |
| `y="y"`        | `y="y"`                              | ✅       |
| `hue="group"`  | `color="group"`                      | ✅（最重要）  |
| `col="panel"`  | `facet_col="panel"`                  | ✅       |
| `size="value"` | `size="value"`                       | ✅       |
| `style="type"` | `symbol="type"` 或 `line_dash="type"` | ⚠️ 名字变了 |
| `col_wrap=3`   | `facet_col_wrap=3`                   | ✅       |


````python
import plotly.express as px


def create_plotly_chart():
    fig = px.bar(x=["A", "B", "C"], y=[1, 3, 2], title="Simple Bar Chart")
    return fig

rx.plotly(
    data=fig,
)
````
