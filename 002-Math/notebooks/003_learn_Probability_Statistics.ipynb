# 2. <a id='toc2_'></a>[概率论](#toc0_)

概率 $P$ 是定义在事件上的一个函数，满足：
1. 非负性: $P(A)≥0$
2. 规范性: $P(Ω)=1$
3. 可列可加性：若事件互不相容，则 $P(\bigcup_{i=1}^\infty A_i)=\sum_{i=1}^\infty P(A_i)$

条件概率与独立性：
- 条件概率: $P(A|B)=\frac{P(A \cap B)}{P(B)}$，表示在 $B$ 发生的前提下 $A$ 发生的概率。
- 乘法公式: $P(A \cap B) = P(A|B) P(B) = P(B|A) P(A)$。
- 全概率公式：若 ${B_i}$ 是完备事件组，则 $P(A) = \sum{_i}{P(A|B_i)P(B_i)}$。
- 贝叶斯公式: $P(B_i|A) = \frac{P(A|B_i)P(B_i)}{\sum{_j}P(A|B_j)P(B_j)}$，是统计推断的基础，将先验知识 $P(B_i)$ 与观测 $A$ 结合得到后验概率。
- 独立事件: $P(A \cap B) = P(A)P(B)$，则称 $A,B$ 独立。独立意味着事件之间无影响。

随机变量及其分布:
- 将随机试验的结果映射为数值，便得到随机变量 $X$。分布描述了 $X$ 取值的概率规律。
- 分布函数: $F(x) = P(X≤x)$，单调不减、右连续、趋于 0/1。
- 离散型: 概率质量函数 $p(x_i) = P(X= x_i)$，如二项分布、泊松分布。
- 连续性: 概率密度函数 $f(x)$, 满足 $P(a≤X≤b) = \int_{a}^{b} f(x) dx$，如正态分布、指数分布、均匀分布。

多维随机变量：
- 联合分布: 描述多个随机变量同时取值的概率；
- 边缘分布: 由联合分布对其它变量求和/积分得到；
- 条件分布: 给定一个变量时，另一个变量的分布；
- 独立性: $F(x, y) = F_X(x) F_Y(y)$ 时，随机变量独立。

数字特征：
- 数学期望 $E[X]$: 随机变量按概率加权的平均值，反映中心位置。
  - 对于离散: $E[X] = \sum{x_i}{p(x_i)}$; 连续: $E[X] = \int{xf{x}dx}$
  - 期望线性: $E[aX = bY] = aE[X] + bE[Y]$
- 方差: $Var(X) = E[(X-μ)^2] = E[X^2] - (E[X])^2$，衡量波动大小。
- 协方差: $Cov(X, Y) = E[ (X-μ_X)(Y-μ_Y) ]$，表示两变量线性相关程度。
- 相关系数: $ρ = \frac{Cov(X, Y)}{\sigma{_X}\sigma{_Y}}$，无量纲，$|ρ|≤1$。
- 矩：高阶矩描述分布形状（偏度、峰度）。

大数定律和中心极限定理 (概率论基石)：
- 大数定律：当试验次数足够多时，样本均值依概率收敛于理论期望。这保证了用频率估计概率的合理性。
- 中心极限定理：大量独立同分布随机变量之和近似服从正态分布。这解释了为什么正态分布在自然界如此普遍，也为统计推断提供了理论依据。



# 3. <a id='toc3_'></a>[数理统计](#toc0_)

数理统计的任务是利用有限的样本数据，对总体分布或其特征 (分布、参数、关系) 进行推断。其逻辑可概括为：总体 → 样本 → 统计量 → 推断。

基本概念：
- 总体：研究对象的全体，通常用一个随机变量 $X$ 表示，其分布未知或部分未知。
- 样本：从总体中随机抽取的 $n$个个体 $(X_i,...X_n)$, 满足独立同分布 (i.i.d.)。
- 统计量：样本的函数，不含未知参数，如样本均值 $\bar{X} = \frac{1}{n} \sum X_{i}, \mathrm{样 本 方 差} S^{2} = \frac{1}{n - 1} \sum \left(X_{i} - \bar{X}\right)^{2}$。
- 推断：基于统计量对总体做出结论，包括参数估计和假设检验两大类。

参数估计， 利用样本信息估计总体分布中的未知参数 $θ$ （如均值、方差）:
- 点估计，构造 $\hat{\theta}$ 作为 $\theta$的估计值。常用方法：
  - 矩估计：用样本矩替换总体矩，解方程得到参数估计。例如，样本均值 $\hat{X}$ 估计总体均值 $μ$。
  - 极大似然估计 (MLE)：选择使样本出现的概率最大的参数值，具有优良性质（相合性、渐近正态性）。
  - 评价标准：无偏性 $(E[\hat\theta = \theta])$、有效性（方差小）、一致性（依概率收敛）。

- 区间估计：点估计无法给出精度，因此给出一个区间 $[L, U]$，使得 $P(L≤θ≤U) = 1 - \alpha$，其中 $1−\alpha$ 为置信水平。常用正态分布、t分布构造置信区间。
  - 常用方法：基于枢轴量，如正态总体均值 $μ$ 的置信区间为 $\bar{X}\pm z_{\alpha/2}\frac{\sigma}{\sqrt{n}}$方差已知）或 t 分布版本（方差未知）。
  - 理解：重复抽样 $100(1−α)%$ 的区间会覆盖真实参数，不是“参数落在区间内的概率”。

假设检验，事先对参数提出一个假设如 $\theta = \theta{_0}$，利用样本判断是否拒绝该假设:
- 基本思想：小概率原理——小概率事件在一次试验中几乎不会发生。若在假设成立下，样本出现极端情况的概率很小，则拒绝假设。
- 两类错误：
  - 第Ⅰ类错误（弃真）：拒绝真实假设，概率记为 $\alpha$（显著性水平）。
  - 第Ⅱ类错误（取伪）：接受错误假设，概率记为 $β$，功效为 $1−β$表示正确拒绝假假设的能力。

- 检验步骤：建立假设（原假设 $H_0$ 与备择假设 $H_1$）→ 选择检验统计量 $T$, 其抽样分布在 $H_0$ 下已知 (或近似) → 计算 p 值（在 $H_0$ 下出现样本结果或更极端结果的概率）→ 若 p < α，拒绝 $H_0$。

- 常用检验：
  - t 检验：检验均值（单样本、两样本、配对）。
  - F 检验：检验方差比。
  - 卡方检验：检验拟合优度、独立性。
  - 非参数检验：不依赖分布假设（如 Wilcoxon 检验）。

回归分析，研究变量间的相关关系，特别是因变量与自变量的依赖关系:
- 线性回归模型：$Y=\beta_0+\beta_1X_1+\cdots+\beta_pX_p+\varepsilon\text{,其中 }\varepsilon\sim N(0,\sigma^2)\mathrm{。}$通过最小二乘法估计参数。
- 最小二乘法：最小化残差平方和得到参数估计 $\hat\beta$，其无偏性和方差可通过样本分布计算。
- 推断：对回归系数进行 t 检验（$β_i = 0$ 与否），整体显著性 F 检验，以及 $R^2$ 衡量拟合优度。
- 模型诊断：残差分析、显著性检验（t检验、F检验）、$R^2$等。

## 4.2. <a id='toc4_2_'></a>[重抽样方法](#toc0_)

在经典统计中，我们通常假设总体服从某种分布（如正态），并基于该假设推导出样本统计量的抽样分布（如 t 分布、F 分布）。但实际中，总体分布可能未知，或者样本量太小无法满足大样本近似。重抽样方法通过从原始样本中反复抽取，生成大量“新样本”，并基于这些新样本计算统计量的分布，从而绕过对总体分布的假设。

重抽样的核心思想：将样本视为“总体”，通过有放回或无放回的重抽样，模拟从总体中抽取多个样本的过程。

### 4.2.1. <a id='toc4_2_1_'></a>[boostrap](#toc0_)

Bootstrap 由 Bradley Efron 在 1979 年提出。其基本思想是：
- 从原始样本 $X_1, X_2, ..., X_n$ 中有放回地抽取 $n$ 个观测值，构成一个 bootstrap 样本（也叫自助样本）。
- 重复上述过程 $B$ 次 (通常1000或更多)，得到 $B$ 个bootstrap样本。
- 在每个bootstrap样本上计算感兴趣的统计量（如均值、中位数、回归系数），得到 $B$ 个bootstrap统计量。
- 这 $B$ 个统计量的分布就近似于改统计量的抽样分布，可以用来估计标准误、构造置信区间、进行假设验证等。

### 4.2.2. <a id='toc4_2_2_'></a>[Jackknife](#toc0_)

- 每次删除一个观测值，用剩余 $n-1$ 个观测值计算统计量，重复 $n$ 次。
- 主要用于估计偏差和标准误，计算量比 bootstrap 小，但适用范围较窄（尤其对非平滑统计量不稳健）。

### 4.2.3. <a id='toc4_2_3_'></a>[Permutation test](#toc0_)

原理：置换检验是一种非参数假设检验方法，不依赖于总体分布。其核心思想是：在原假设下，样本的组别标签可以随机置换（即观测值与组别无关），从而构建检验统计量的抽样分布。

步骤：
1. 计算原始样本的检验统计量（如两组的均值差 $T_obs$）。
2. 在原假设成立的条件下，将组别标签随机置换（打乱），重新计算置换后的统计量 ${T^{*}}$。
3. 重复置换 $B$ 次（通常1000或更多），得到 $B$ 个置换统计量，形成原假设下的经验分布。

## 4.3. <a id='toc4_3_'></a>[贝叶斯学派](#toc0_)

贝叶斯定理描述了条件概率之间的逆关系。对于两个事件 $A$ 和 $B$ 有：

$$P (A | B) = \frac{P (B | A) P (A)}{P (B)}$$

其中：
- $P(A)$是先验概率 (prior), 表示在观测数据前面对 $A$ 的信念；
- $P(B|A)$ 是似然 (likelihood), 表示在 $A$ 成立时 $B$ 出现的概率；
- $P(B)$ 是边缘概率, 通常作为归一常数；
- $P(A|B)$ 是后验概率 (posterior), 表示观测到 $B$ 后对 $A$ 的更新信念。
