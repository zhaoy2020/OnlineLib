# 2. <a id='toc2_'></a>[极限](#toc0_)

```python
def show(x, y, text, ax=None):
    if ax is None:
        ax = plt.gca()
    ax.plot(x, y)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.text(0.1, 0.9, s=f'${text}$', fontsize=7, ha='left', va='top', transform=ax.transAxes)


x = np.linspace(-10, 10, 100)
y = 1 / x
y1 = np.sin(x) / x

fig, axs = plt.subplots(1, 2, figsize=(6, 3))
show(x, y, 'y = 1/x', ax=axs[0])
show(x, y1, 'y = sin(x)/x', ax=axs[1])
x = sp.symbols('x')
f = sp.sin(x) / x
limit_val = sp.limit(f, x, 0)
print(f'当x趋近于0时，sin(x)/x的极限值为：{limit_val}')
```

```python
def func(x):
    a = np.sin(x) / x
    b = np.cos(x) / x
    z = np.sqrt(a**2 + b**2)
    return z

x = np.linspace(0, 10, 100)
y = func(x)
show(x, y, 'z = sqrt((sin(x)/x)^2 + (cos(x)/x)^2)')

x = sp.symbols('x')
f = sp.sqrt((sp.sin(x)/x)**2 + (sp.cos(x)/x)**2)
limit_val = sp.limit(f, x, 0)
print(f'当x趋近于0时，z的极限值为：{limit_val}')
```

# 3. <a id='toc3_'></a>[导数](#toc0_)

差分极限。

```python
f = lambda x: x**2

x0 = 1.0
h = 1e-5
deriv_approx = (f(x0 + h) - f(x0)) / h
print(f'在x={x0}处，sin(x)/x的导数近似值为：{deriv_approx}')


x = sp.Symbol('x')
f = x**2
df = sp.diff(f, x)
print(f'函数f(x) = {f}的导数为：{df}')
```
